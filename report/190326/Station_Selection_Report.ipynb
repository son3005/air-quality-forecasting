# 🗺️ Station Selection Report — Bản Đồ Bán Kính 100km
Mỗi trạm được vẽ một **vòng tròn bán kính 100km**. Các vòng tròn chồng lấp nhau cho thấy trạm nào có thể liên kết Graph.  
- 🔴 **Đỏ:** Cụm Bắc Bộ đề xuất  
- 🔵 **Xanh:** Cụm Nam Bộ đề xuất  
- ⚪ **Xám:** Chưa chọn / Đang cân nhắc

**Nguồn dữ liệu giao thông:**  
- Hà Nội: >8 triệu phương tiện (VnExpress, 01/2024)  
- TP.HCM: >9 triệu phương tiện (Tuổi Trẻ, 09/2023)


```python

import pandas as pd
import folium

data = [
    (1, 'Hà Nội', 'Cầu Giấy', 21.0323, 105.8007, 'North'),
    (2, 'Hà Nội', 'Thanh Xuân', 20.9947, 105.7998, 'North'),
    (3, 'Hà Nội', 'Tây Hồ', 21.0690, 105.8105, 'North'),
    (4, 'Thanh Hóa', 'Thọ Xuân', 19.9500, 105.5000, 'North'),
    (5, 'Hà Nội', 'Gia Lâm', 21.0535, 106.0071, 'North'),
    (6, 'Hà Nội', 'Ba Vì', 21.0833, 105.3833, 'North'),
    (7, 'TP.HCM', 'Quận 1', 10.7807, 106.6994, 'South'),
    (8, 'TP.HCM', 'Quận 3', 10.7749, 106.6863, 'South'),
    (9, 'TP.HCM', 'Bình Thạnh', 10.8033, 106.6967, 'South'),
    (10, 'TP.HCM', 'Quận 10', 10.7682, 106.6663, 'South'),
    (11, 'TP.HCM', 'Quận 11', 10.7638, 106.6436, 'South'),
    (12, 'TP.HCM', 'Bình Chánh', 10.6954, 106.5913, 'South'),
    (13, 'Đà Nẵng', 'Thanh Khê', 16.0706, 108.1910, 'Drop'),
    (14, 'Đà Nẵng', 'Sơn Trà', 16.0607, 108.2326, 'Drop'),
    (15, 'Đà Nẵng', 'Liên Chiểu', 16.0177, 108.2038, 'Drop'),
    (16, 'Hải Phòng', 'Đồ Sơn', 20.7136, 106.7894, 'North'),
    (17, 'Hải Phòng', 'Hồng Bàng', 20.8607, 106.6790, 'North'),
    (18, 'Cần Thơ', 'Ninh Kiều', 10.0371, 105.7883, 'Drop'),
    (19, 'Cần Thơ', 'Cái Răng', 10.0009, 105.7510, 'Drop'),
    (20, 'Cần Thơ', 'Bình Thủy', 10.0743, 105.7397, 'Drop'),
    (21, 'Khánh Hòa', 'Cam Ranh', 11.9214, 109.1591, 'Drop'),
    (22, 'Khánh Hòa', 'Nha Trang', 12.2451, 109.1943, 'Drop'),
    (23, 'Khánh Hòa', 'Diên Khánh', 12.2549, 109.0933, 'Drop'),
    (24, 'Đồng Nai', 'Biên Hòa', 10.9447, 106.8243, 'South'),
    (25, 'Đồng Nai', 'Long Thành', 10.7891, 106.9503, 'South'),
    (26, 'Đồng Nai', 'Nhơn Trạch', 10.7229, 106.8834, 'South'),
    (27, 'Nam Định', 'TP. Nam Định', 20.4339, 106.1773, 'North'),
    (28, 'Thanh Hóa', 'TP. Thanh Hóa', 19.8000, 105.7667, 'North'),
    (29, 'Nghệ An', 'TP. Vinh', 18.6734, 105.6923, 'Drop'),
    (30, 'Vĩnh Long', 'TP. Vinh Long', 10.2537, 105.9722, 'Drop'),
    (31, 'Trà Vinh', 'TP. Trà Vinh', 9.9472, 106.3423, 'Drop'),
    (32, 'An Giang', 'Rạch Giá', 10.0124, 105.0809, 'Drop')
]

df = pd.DataFrame(data, columns=['ID', 'Province', 'District', 'Lat', 'Lon', 'Cluster'])

# -- Create map --
m = folium.Map(location=[15.5, 107.0], zoom_start=6, tiles='CartoDB positron')

colors = {'North': 'red', 'South': 'blue', 'Drop': 'gray'}
fill_colors = {'North': '#ff4444', 'South': '#4444ff', 'Drop': '#999999'}

for _, row in df.iterrows():
    color = colors[row['Cluster']]
    fill = fill_colors[row['Cluster']]
    
    # Vòng tròn bán kính 100km cho MỖI trạm
    folium.Circle(
        location=(row['Lat'], row['Lon']),
        radius=100000,  # 100km   
        color=color,
        fill=True,
        fill_color=fill,
        fill_opacity=0.08,
        weight=1.5,
        popup=f"Bán kính 100km — Trạm {row['ID']}"
    ).add_to(m)
    
    # Điểm trạm (marker nhỏ)
    folium.CircleMarker(
        location=(row['Lat'], row['Lon']),
        radius=6,
        popup=f"<b>Trạm {row['ID']}</b><br>{row['District']}, {row['Province']}<br>Cluster: {row['Cluster']}",
        color='black',
        weight=1,
        fill=True,
        fill_color=fill,
        fill_opacity=0.9
    ).add_to(m)
    
    # Label trạm
    folium.Marker(
        location=(row['Lat'], row['Lon']),
        icon=folium.DivIcon(html=f'<div style="font-size:9px; font-weight:bold; color:{color}; text-shadow: 1px 1px white;">{row["ID"]}</div>')
    ).add_to(m)

# Legend
legend_html = """
<div style="position: fixed; bottom: 30px; left: 30px; z-index: 1000; 
     background: white; padding: 12px; border-radius: 8px; border: 2px solid #333; font-size: 13px;">
<b>Chú thích</b><br>
🔴 Cụm Bắc Bộ (đề xuất)<br>
🔵 Cụm Nam Bộ (đề xuất)<br>
⚪ Chưa chọn<br>
⭕ Bán kính 100km mỗi trạm
</div>
"""
m.get_root().html.add_child(folium.Element(legend_html))

display(m)
```

## Hướng Dẫn Đọc Bản Đồ
- **Zoom vào từng vùng** để xem các vòng tròn 100km chồng lấp.
- Trạm nào có vòng tròn **chồng lấp nhiều** với các trạm khác → Liên kết Graph mạnh.
- Trạm nào **cô lập** (vòng tròn không chạm ai) → Ứng cử viên bị loại.
- Tại cụm HCM: Các vòng xanh chồng kín lên nhau → Xác nhận data trùng lặp (r=0.999).
- Tại cụm Bắc: Vòng đỏ phủ đều → Mỗi trạm đóng góp thông tin riêng biệt.
